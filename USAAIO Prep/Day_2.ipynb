```python
# Exercise 1
def normalize(X):
  min = X.min(axis=0)
  max = X.max(axis=0)
  print(f"Min: {min}, Max: {max}")
  return (X-min) / (max-min)
test_array = np.random.randn(6,5)
print(test_array)
normalized = normalize(test_array)
print(f"Normalized: {normalized} with shape {normalized.shape}")
```

```python
# Exercise #2
def normVerifier(X_norm):
  min = X_norm.min(axis=0)
  max = X_norm.max(axis=0)
  for x in min:
    if not x>=0:
      return np.where(min == x)
    else:
      return None
  for x in max:
    if not x<=1:
      return np.where(max == x)
    else:
      return None
#Fail
print(normVerifier(test_array))
#Pass
print(normVerifier(normalized))
```

```python
# Exercise #3
'''
Given X with shape (N, D):
Compute an (N, N) matrix where entry (i, j) is the Euclidean distance between X[i] and X[j].
Confirm:
The diagonal entries are all zero
The matrix is symmetric
No explicit loops

No scipy

Use broadcasting and matrix multiplication'''
X = np.random.rand(4, 2)
print(X)
def coolfunc(X):
  sq_norms = np.sum(X**2, axis=1, keepdims=True)
  print(f"square norms: {sq_norms}")
  dist_sq = sq_norms - 2 * X @ X.T + sq_norms.T
  print(f"distance norms: {dist_sq}")
  return np.sqrt(dist_sq)

print(coolfunc(X))
```
