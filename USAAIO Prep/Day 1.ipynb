```python
import numpy as np
x = np.array([1, 2, 3, 4])
print(x)
#Part 1: Convert to shape (4, 1)
x = x.reshape(4, 1)
print(x)
#Part 2: Convert to shape (1, 4)
x = x.reshape(1, 4)
print(x)
```

The difference between shapes (4,) (4, 1) and (1, 4) is that 4 is just a single row of numbers, (4, 1) is a 4x1 matrix, while (1,4) is a 1x4 matrix. In a way, the outputs of (4,) and (1, 4) are similar.
