# Linear Regressions with Linear Algebra
The following notebook explores Linear Regressions but instead of using a Gradient Descent Algorithm, it relies on Linear Algebra concepts such as Gaussian Elimination and Projections.

This serves more as a proof of concept than an actual practical use case.

The LinearRegression class below is trying to find a solution to the linear system $Xw=y$ where $X$ and $y$ and matricies and $w$ is a column vector for weights.

One major difference as compared to the Gradient Descent algorithm is that Linear Algebra solves the weights immediately, rather than Gradient Descent which starts with random values and adjusts the weights to try to converge towards the actual weights and biases.

In essence, this is not really a "Machine Learning" sort of thing where the machine tries to find the pattern of data, rather a caveat of Linear Algebra.
Interestingly scikitlearn uses this "under the hood" for some of its smaller linear regressions.

## How the Model Works
The following Linear Regression model has the following setup.
Given data in the form of lists
$$\begin{bmatrix} X_0 \\ X_1 \\ X_2 \\ ... \\ X_n \end{bmatrix}$$, we want to find data in the form of $$\begin{bmatrix} y_0 \\ y_1 \\ y_2 \\ ... \\ y_n \end{bmatrix}$$.
In order to achieve this, we must multiply matrix $X$ by a vector $w$, consisting of the weights and biases.
Our final matrix equation setup will look something like this:
$$\begin{bmatrix} X_0 & 1 \\ X_1 & 1 \\ X_2 & 1 \\ ... & ... \\ X_n & 1 \end{bmatrix} \cdot \begin{bmatrix} w \\ b \end{bmatrix} = \begin{bmatrix} y_0 \\ y_1 \\ y_2 \\ ... \\ y_n \end{bmatrix}$$

Notice how this accounts for a biase $b$, and a weight $w$ that applies to each value for feature $X$.

```python
# transposing algorithm
import numpy as np
'''Transposing is heavily used throught this process, and therefore I will be using one of my own transposing algorithms.'''
def transpose(X):
  X = np.array(X)
  columns = len(X[1])-1
  X_tranpose = []
  for i in range(0, columns+1):
    columntoRow = X[:, i]
    X_tranpose.append(columntoRow)

  return np.array(X_tranpose)


matrix = np.random.random([2, 3])
transposed = transpose(matrix)
print(f"Original matrix: {matrix} \n Tranposed Matrix: {transposed}")
```

```python
import numpy as np
class LinearRegression():
  def __init__(self):
    pass

  def fit(self, X, y):
    ## append columns of np.ones to include biases
    good_X = []
    for row in X:
      row.append(1)
      good_X.append(row)
    X = good_X
    self.good_X = good_X
    transposed_X = transpose(X=X)
    ## transformed equation X^T * X * w = X^T * y
    # Building augmented matrix
    left_side = np.matmul(transposed_X, X)
    right_side = np.matmul(transposed_X, y)

    # Perform Gaussian Elimination
    ## Given its a linear regression it will merely by 2x2

    multiplier = left_side[1,0] / left_side [0, 0]
    row_1L = left_side[0,:]
    row_1R = right_side[0]
    row_2L = left_side[1,:]
    row_2R = right_side[1]
    new_row2L = row_2L - (multiplier * row_1L)
    new_row2R = row_2R - (multiplier * row_1R)
    left_side[1, :] = new_row2L
    right_side[1] = new_row2R

    #back substituiton
    ## bias
    self.bias = right_side[1] / left_side [1, 1]
    self.weight = (right_side[0] - (left_side[0, 1] * self.bias))/left_side[0,0]


  def predict(self, X):
    good_X = []
    for row in X:
        row.append(1)
        good_X.append(row)
    return np.matmul(good_X, [[self.weight], [self.bias]])


```

```python
# Testing
Xdata = np.random.random(size=(100, 1))  # already column format, no transpose needed
ydata =

3 * Xdata.flatten() + 1          # flatten to get a clean 1D array

regress = LinearRegression()
regress.fit(Xdata.tolist(), ydata.tolist())  # fit the model
predictions = regress.predict(Xdata.tolist())

print("Weight:", regress.weight)  # should be close to [3, 1]
print("Bias: ", regress.bias)
```

The above algorithm performs on linear data. The beauty of matricies and gaussian elimination is that this will work for matricies of larger dimensions. Thus, we experiment with matricies in higher dimensions and datasets with more features.

In order to reduce the complexity (having to Gaussian Eliminiate on a big matrix would be a pain to write the code for) - we will first apply Gram-Schmidt to convert $X$ to $QR$, before projecting $y$ onto the new matrix.

```python
# Convert A to QR
import numpy as np
class LinearRegression2():
  def __init__(self):
    pass

  def fit(self, X, y):
    # convert X to orthonormal Q and upper triangular R
    transposedX = transpose(X)
    gramX = np.array(transposedX[0])
    gramX = gramX.reshape(len(X), 1)
    # Gram-Schmidt to find Q
    for col in transposedX[-(len(X[0])-1):]:
      column = col
      for i in range(0, (transposedX == col).all(axis=0).argmax()):
        newCol = column - (np.matmul(gramX[:, i], column)/np.matmul(gramX[:, i], X[:, i]))
        column = newCol
      gramX = np.column_stack([gramX, column])
    #schmidt part
    for i in range(0, len(gramX[0])):
      length = 0
      for j in gramX[:, i]:
        length+=j
      gramX[:, i] = gramX[:, i] / length
    self.Q = gramX
    self.R = np.matmul(transpose(self.Q), X)

    ## solve regression by projecting b onto Q
    # QRx = b --> Rx = Q^Tb
    right_side = np.matmul(transpose(self.Q), y)

    #back-substitution
    x = len(self.R[0])-2
    self.weights = [right_side[-1]/self.R[-1, -1]]
    self.weights = np.array(self.weights)
    #self.weights.reshape(len(self.R[0]), 1)
    left_side = self.R
    left_side[:, -1] = self.weights[0] * self.R[:, -1]
    for i in range(2, len(X[0])+1):
      length = 0
      for j in range(0, len(X[0])):
        if j!=i:
          length+=left_side[-i, j]
      weight = [(right_side[-i] - length)/left_side[-i, -i]]
      weight = np.array(weight)
      self.weights = np.insert(self.weights, 0, weight)


  def predict(self, X):
    return np.matmul(X, self.weights)




```

```python
Xdata = np.random.random([100, 5])
ydata  = 3.28 * Xdata[:, 0] + 4.12089 * Xdata[:, 1] + 5.238 * Xdata[:, 2] + 7.293874 * Xdata[:, 3] + 1.1092 * Xdata[:, 4]
```

```python

regression = LinearRegression2()
regression.fit(Xdata, ydata)

print(f"Weights: {regression.weights}")
print(f"Predicted: {regression.predict(Xdata)}")
print(f"Actual: {ydata}")
```

Alternatively, we can directly solve this by applying the big formula to solve for $\hat{x}$ via projecting $b$ onto $A$.
Here is a quick derivation of the formula:

When $b$ is projected onto a matrix $A$, the error vector is perpendicular to that matrix
Using that, we know $$A^T(b - Ax) = 0$$
$$A^Tb = A^TAx$$ Now we must isolate $x$
$$x = (A^TA)^{-1} A^Tb$$

Keep in mind we are using $X$ and $y$ instead of $A$ and $b$ for our variables and $w$ for weights (bad naming I can't lie).



```python
# Define Function
import numpy as np
class LinearRegression3():
  def __init__(self):
    pass

  def fit(self, X, y):
    # solve for weights using projection formula
    self.weights = np.matmul(np.linalg.inv(np.matmul(transpose(X), X)), np.matmul(transpose(X),y))

  def predict(self, X):
    return np.matmul(X, self.weights)

```

```python
# Testing
Xdata = np.random.random([100, 5])
ydata  = 3.28 * Xdata[:, 0] + 4.12089 * Xdata[:, 1] + 5.238 * Xdata[:, 2] + 7.293874 * Xdata[:, 3] + 1.1092 * Xdata[:, 4]

regression = LinearRegression3()
regression.fit(Xdata, ydata)
predictions = regression.predict(Xdata)
print(f"Weights: {regression.weights}")
print(f"Predictions: {predictions} Actual: {ydata}")
```
