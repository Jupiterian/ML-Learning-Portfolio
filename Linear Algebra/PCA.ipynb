# Principal Component Analysis (PCA)
A PCA algorithm written from scratch (based on numpy).

```python
import numpy as np
```

```python
class PCA():
  def __init__(self, components):
    self.components = components

  def fit(self, A):
    # Step 1: center matrix A -> B
    B = A.copy()
    for i in range(0, B.shape[1]):
      # find mean
      sum = 0
      for x in B[:, i]:
        sum+=x
      mean = sum/B.shape[0]

      # center column
      B[:, i] = B[:, i] - mean

    # compute svd of B
    SVD = np.linalg.svd(B)

    # parse the first n components
    U = SVD.U[:, :self.components]
    S = np.diag(SVD.S)
    S = S[:self.components, :self.components]
    Vh = SVD.Vh[:self.components, :]
    self.reconstructed = np.matmul(np.matmul(U,S), Vh)
```

```python
X = np.random.random([1000, 6])
PCA = PCA(components=4)
PCA.fit(X)
PCA.transform()
```
