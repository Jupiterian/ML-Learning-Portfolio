## Transpose Algorithms
Given a matrix $A$, return matrix $A^T$.

Solution 1 turns columns of $A$ into rows of $A^T$.
Solution 2 turns rows of $A$ into columns of $A^T$. Because numpy arrays are defined in terms of rows, a template matrix is created and then each element in each row is filled in this template matrix for Solution 2.

```python
import numpy as np
def transposer(A):
  # Columns of A -> Rows of A^T

  ## Get matrix shape in numpy indicies
  rows = len(A)-1
  columns = len(A[1])-1

  A_transpose = []
  # Get columns and convert to rows
  for x in range (0, columns+1):
    column = A[:, x]
    A_transpose.append(column)

  return np.array(A_transpose)


# Test
matrix = np.random.random(size=[3, 3])
print(f"Matrix: {matrix}")
print(f"Tranpose: {transposer(matrix)}")
```

```python
# Note: Used small hint from AI to get the idea of zero matricies
import numpy as np
def transposer(A):
  # Columns of A -> Rows of A^T

  ## Get matrix shape in numpy indicies
  rows = len(A)-1
  columns = len(A[1])-1

  # Get columns and convert to rows
  zeroMatrix = np.random.random(size=[columns+1, rows+1])
  for y in range (0, rows+1):
    row = A[y, :]
    indicie = 0
    for x in row:
      zeroMatrix[indicie, y] = x
      indicie+=1

  return np.array(zeroMatrix)


# Test
matrix = np.random.random(size=[3, 3])
print(f"Matrix: {matrix}")
print(f"Tranpose: {transposer(matrix)}")
```
