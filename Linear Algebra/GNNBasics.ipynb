# GNN Basics
The following notebook implements the basics/core funcitionality of a GNN manually that turns a graph into data. This aims at recreating a single layer of a GCN. In the future I plan to add more to make this a full layered GCN. In full disclosure, much of this is also simplified compared to an actual GCN.

Its important to note that in the real world data has to be adjusted in order to create the $A$ and $X$ matricies of a GCN, but here for simplicity $A$ and $X$ are pregenerated in the ```input``` function.

````python
import numpy as np
class GCNLayer():
  def genInput(self, nodes):
    self.A = np.random.choice([-1, 1], size=[nodes, nodes])
    self.X = np.random.random([nodes, nodes]) # ensures that no prior knowledge or weightage on nodes is set
````
