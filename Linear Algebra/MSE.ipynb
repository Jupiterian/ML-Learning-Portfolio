# Mean Squared Error
The following notebook explores Mean-Squared Error, a loss function used often in Neural Networks. Currently, given my lack of multivariable calc knowledge it only calculates the loss and passes it off. In the future I plan to add an 'optimizer' to update the weights.

```python
class MSE():
  def lossValue(self, y_pred, y_actual):
    lossVals = []
    for i in range(0, len(y_pred)):
      lossVals.append((y_actual[i] - y_pred[i])**2)
    #calculate average
    sum = 0
    for x in lossVals:
      sum+=x
    mse=sum/len(y_pred)
    return mse
```

## Vectorized MSE
Through Linear Algebra, instead of iterating through the lists and calculating the loss for each specific element, we can use vectorization to simplify the MSE algorithm.


```python
class MSE2():
  def lossValue(self, y_pred, y_actual):
    squared_errors = (y_actual - y_pred)**2
    sse = 0
    #np.sum essentially works this way as well so there's no point
    for x in squared_errors:
      sse+=x
    mse = sse/len(y_actual)
    return mse
```

## Testing

```python
import numpy as np
actual = np.random.random([100, 1])
pred = np.random.random([100,1])
loss1 = MSE()
loss2 = MSE2()
print(loss1.lossValue(pred, actual))
print(loss2.lossValue(pred, actual))
```
