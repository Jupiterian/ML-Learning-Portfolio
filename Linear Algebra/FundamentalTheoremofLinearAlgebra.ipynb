# Fundamental Theorem of Linear Algebra

The Fundamental Theorem of Linear Algebra defines the four fundamental subspaces: column spaces, row spaces, nullspaces, and left nullspaces. This notebook aims to define and find the dimension of each of these subspaces, provided a matrix.

## Definiton
The Fundamental Theorem of Linear Algebra states that given an matrix $A$ of shape $m × n$:
 - The Column Space of $A$ is defined in $\mathbb{R}^m$ and has a dimension of the matrix rank $r$.
 - The Row Space of $A$ (also the columns pace of $A^T$ is) is defined in $\mathbb{R}^n$ and has a dimension of $r$.
 - The Nullspace of $A$ is defined in $\mathbb{R}^n$ and has a dimension of $n-r$.
 - The left Nullspace of $A$ is defined in $\mathbb{R}^m$ and has a dimension of $m-r$.

 Additionally, the row space is orthogonal to the nullspace and the column space is orthogonal to the left nullspace.
