# Gradient Descent
The following notebook explores that Gradient Descent algorithm through both a visual and mathematical approach. This is built entirely from scratch. For simplicity it minimizes the function $f(x) = x^2 + y^2$.

```python
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.animation import FuncAnimation
import math
```

```python
def GradientDescent(lr, coords):
  history = []
  while math.sqrt(coords[0]**2 + coords[1]**2) > 0.1:
    coords = coords - (2 * lr * coords)
    history.append([coords[0], coords[1]])
  return history
```

```python
rng = np.random.default_rng()
coordinates = rng.integers(low=0, high = 1000, size = 2)
history = GradientDescent(0.1, coordinates)
print(history)
```
